```python
import urllib.request as urllib
def upload_to_volume(url: str, file_names: list[str], distination_volume: str):
    
    #url = "https://raw.githubusercontent.com/kuljotSB/DatabricksUdemyCourse/refs/heads/main/"
    for file in file_names:
        urllib.urlretrieve(f"{url}/{file}", f"{distination_volume}/{file}")
    
    print(f"Files uploaded to {distination_volume}")
```
