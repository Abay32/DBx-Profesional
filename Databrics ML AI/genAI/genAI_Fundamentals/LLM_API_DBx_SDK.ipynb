```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service.serving import ChatMessage, ChatMessageRole

# Create a workspace client
w = WorkspaceClient()
```

```python
# Sending a chat completion request to the LLM
response = w.serving_endpoints.query(
    name='databricks-claude-sonnet-4-6',
    messages=[
        ChatMessage(role=ChatMessageRole.SYSTEM, content="አንተ ባትማን ነህ — የባህር ዳር ጠባቂ።"),
        ChatMessage(role=ChatMessageRole.USER, content="ባህር ዳር ዛሬ እንዴት ናት?")
    ],
    max_tokens=1024
)

print(f" RESPONSE: \n{response.choices[0].message.content}")
print(f" COST: {response.usage.total_tokens} tokens") 
```
