```python
from pyspark.sql import functions as F
division_key_value = {
    "Energy Bars": "Nutrition Bars",
    "Protein Bars": "Nutrition Bars",
    "Granola &amp; Cereals": "Breakfast Food",
    "Recovery Dairy": "Dairy &amp; Recovery",
    "Health Snacks": "Health Snacks",
    "Electrolyte Mix": "Hydration &amp; Electrolytes"
}

# Convert dict → Spark map expression
mapping_expr = F.create_map(
    *[F.lit(x) for kv in division_key_value.items() for x in kv]
)
print(mapping_expr)

for kv in division_key_value.items():
  print(kv)
```

```python

division_map = F.create_map(
    F.lit("Energy Bars"), F.lit("Nutrition Bars"),
    F.lit("Protein Bars"), F.lit("Nutrition Bars"),
    F.lit("Granola &amp; Cereals"), F.lit("Breakfast Food"),
    F.lit("Recovery Dairy"), F.lit("Dairy &amp; Recovery"),
    F.lit("Health Snacks"), F.lit("Health Snacks"),
    F.lit("Electrolyte Mix"), F.lit("Hydration &amp; Electrolytes")
)
print(division_map)
```
